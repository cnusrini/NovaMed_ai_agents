# NovaMed — A LangGraph based AI Agents Built on One Drug's Journey

Every example below is a subsystem of **NovaMed(A fictiosious drug manufacturer)**: an agentic system that
takes a drug batch from raw-material intake through formulation, quality
control, regulatory documentation, and market-facing drug information, with
a human pharmacist able to intervene at any critical checkpoint.

## The Drug Manufacturing Journey — where each agent fits

```
RAW MATERIAL INTAKE  →  FORMULATION  →  QUALITY CONTROL
Ch.1 Batch Tracker       Ch.1 Batch        Ch.3 Batch Release Agent
Ch.7 Inventory Agent     Tracker           (purity + dosage variance)
                                                 │
                                     ┌───────────┴───────────┐
                                   PASS                     FAIL
                                     │                       │
                          REGULATORY REVIEW          DEVIATION HANDLING
                          Ch.4 Supervisor →           Ch.8 Deviation Report
                          regulatory_agent             Agent (generate →
                                     │                  QA review → revise)
                          BATCH DOCUMENTATION                │
                          Ch.4 Supervisor →           back to FORMULATION
                          documentation_agent
                                     │
                          HUMAN PHARMACIST SIGN-OFF   ← agents inform,
                                     │                   human decides
                          MARKET-FACING DRUG INFO
                          Ch.6 Drug Info & Pricing Assistant
                                     │
                          POST-MARKET SURVEILLANCE
                          Ch.10 Adverse Event Screener
                                     │
                          ONGOING COMPLIANCE
                          Ch.11 Regulatory Guidance Summarizer

R&D (Ch.5) feeds new/reformulated compounds INTO raw material intake.
SOP Assistant (Ch.2) and Plant Operations Router (Ch.9) run alongside
every stage above, supporting plant-floor operators directly.
```

## How to run this notebook

1. Run the **Setup** cell below to install dependencies.
2. Run the **Config** cell and choose `openai` or `ollama` as your provider.
   - `ollama` needs zero API key — install [Ollama](https://ollama.com/download),
     then `ollama pull llama3.2` and `ollama pull nomic-embed-text` before running.
   - `openai` needs `OPENAI_API_KEY` set as an environment variable.
3. Run cells top to bottom — each example builds on state/tools/agents from
   earlier cells, so later cells assume earlier ones have already run.


## Setup

Installs everything needed for every example in this notebook.

## Config — choose your LLM provider

Set `LLM_PROVIDER` to `"ollama"` (free, local, no API key) or `"openai"`
(needs `OPENAI_API_KEY`). Every example below calls `get_llm()` /
`get_embeddings()` so this is the only place you need to change providers.

````python
import os

LLM_PROVIDER = "ollama"   # "ollama" or "openai"
OLLAMA_MODEL = "llama3.2"
OLLAMA_EMBED_MODEL = "nomic-embed-text"
OPENAI_MODEL = "gpt-4o-mini"

def get_llm(temperature: float = 0):
    """Returns the chat model configured by LLM_PROVIDER above."""
    if LLM_PROVIDER == "ollama":
        from langchain_ollama import ChatOllama
        return ChatOllama(model=OLLAMA_MODEL, temperature=temperature)
    else:
        from langchain_openai import ChatOpenAI
        return ChatOpenAI(model=OPENAI_MODEL, temperature=temperature)

def get_embeddings():
    """Returns the embedding model configured by LLM_PROVIDER above."""
    if LLM_PROVIDER == "ollama":
        from langchain_ollama import OllamaEmbeddings
        return OllamaEmbeddings(model=OLLAMA_EMBED_MODEL)
    else:
        from langchain_openai import OpenAIEmbeddings
        return OpenAIEmbeddings()

print(f"Provider: {LLM_PROVIDER}")
````

# Example 1 — Batch Stage Tracker

**Use case:** Before PharmaOps can make any decision about a batch, it needs
a record of what's happened to that batch so far. This is the simplest
possible LangGraph — no LLM, no decisions — but it establishes the state
shape (`batch_id`, `stage_log`) that every later example reads from or
writes to. This is the plumbing lesson: state flows through nodes, edges
carry control not data.

````python
from typing import TypedDict
from langgraph.graph import StateGraph, START, END

class BatchTrackingState(TypedDict):

    batch_id: str
    stage_log: str  # human-readable trail of what's happened to this batch

def log_material_receipt(state: BatchTrackingState) -> dict:

    """Record that raw materials for this batch have arrived and been logged."""
    updated_log = state["stage_log"] + " -> Raw materials received"

    print(f"[MATERIAL RECEIPT] {updated_log}")
    return {"stage_log": updated_log}

def log_formulation_complete(state: BatchTrackingState) -> dict:

    """Record that the batch has been formulated per the master batch record."""
    updated_log = state["stage_log"] + " -> Formulation complete"

    print(f"[FORMULATION] {updated_log}")
    return {"stage_log": updated_log}

def log_qc_check(state: BatchTrackingState) -> dict:

    """Record that the batch has passed through the quality control checkpoint."""
    updated_log = state["stage_log"] + " -> QC checkpoint passed"
    
    print(f"[QUALITY CONTROL] {updated_log}")
    return {"stage_log": updated_log}

batch_graph = StateGraph(BatchTrackingState)
batch_graph.add_node("log_material_receipt", log_material_receipt)
batch_graph.add_node("log_formulation_complete", log_formulation_complete)
batch_graph.add_node("log_qc_check", log_qc_check)

batch_graph.add_edge(START, "log_material_receipt")
batch_graph.add_edge("log_material_receipt", "log_formulation_complete")
batch_graph.add_edge("log_formulation_complete", "log_qc_check")
batch_graph.add_edge("log_qc_check", END)

batch_tracker_app = batch_graph.compile()

initial_batch_state = {"batch_id": "PC-2026-0091", "stage_log": "Batch opened"}
final_batch_state = batch_tracker_app.invoke(initial_batch_state)
print()
print(final_batch_state)
````

# Example 2 — SOP Assistant

**Use case:** Plant operators constantly need to check Standard Operating
Procedures — gowning protocol, deviation-reporting rules, cleanroom entry
steps — without pulling a supervisor off the floor. This is the first
LLM-backed component in PharmaOps and the first place conversation memory
matters: an operator's follow-up question only makes sense with prior turns
in context.

````python
from langgraph.graph import StateGraph, MessagesState, START, END
from langchain_core.messages import HumanMessage, SystemMessage

sop_assistant_llm = get_llm(temperature=0.2)

def answer_sop_question(state: MessagesState) -> dict:
    """
    Reads the operator's conversation so far, answers against SOP knowledge,
    and returns the response to be appended to the conversation.
    """
    sop_assistant_persona = SystemMessage(content="""
    You are the PharmaCore SOP Assistant. You help plant operators understand
    standard operating procedures for batch handling, cleanroom protocol, and
    GMP documentation requirements. Be precise, and say so if a procedure
    isn't in your knowledge rather than guessing.
    """)
    full_conversation = [sop_assistant_persona] + state["messages"]
    operator_response = sop_assistant_llm.invoke(full_conversation)
    return {"messages": [operator_response]}

sop_graph = StateGraph(MessagesState)
sop_graph.add_node("answer_sop_question", answer_sop_question)
sop_graph.add_edge(START, "answer_sop_question")
sop_graph.add_edge("answer_sop_question", END)

sop_assistant_app = sop_graph.compile()
````

````python
# Scripted conversation (swap for your own questions, or wrap in a loop)
conversation_history = []

def ask_sop_assistant(question: str):
    global conversation_history
    conversation_history.append(HumanMessage(content=question))
    result = sop_assistant_app.invoke({"messages": conversation_history})
    conversation_history = result["messages"]
    print(f"Operator: {question}")
    print(f"SOP Assistant: {conversation_history[-1].content}\n")

ask_sop_assistant("What's the gowning procedure before entering a Class B cleanroom?")
ask_sop_assistant("And how often does that gown need to be changed?")
````

# Example 3 — Batch Release Agent

**Use case:** A pharmacist asks "can this batch be released?" This requires
*acting*, not just answering — the agent must pull lab purity data and
check dosage math before it can respond. First ReAct loop: reason -> call a
tool -> observe -> reason again -> answer.

````python
from langchain_core.tools import tool
from langgraph.prebuilt import ToolNode, tools_condition

@tool
def check_batch_purity(batch_id: str) -> str:
    """Look up lab purity test results for a given batch ID."""
    lab_results = {
        "PC-2026-0091": "Purity: 99.6% | Impurity profile: within spec | Status: PASS",
        "PC-2026-0044": "Purity: 96.1% | Impurity profile: exceeds threshold | Status: FAIL",
    }
    return lab_results.get(batch_id, f"No lab record found for batch {batch_id}")

@tool
def calculate_dosage_variance(target_mg: float, measured_mg: float) -> str:
    """Calculate the % variance between target and measured dosage per unit."""
    variance_pct = abs(measured_mg - target_mg) / target_mg * 100
    verdict = "within USP tolerance" if variance_pct <= 5 else "OUT OF TOLERANCE"
    return f"Variance: {variance_pct:.2f}% - {verdict}"

@tool
def lookup_drug_monograph(drug_name: str) -> str:
    """Retrieve the regulatory monograph summary for a drug (dosage form, storage, indications)."""
    monographs = {
        "metformin": "Oral biguanide, 500-1000mg tablets, store below 25C, indicated for T2DM.",
    }
    return monographs.get(drug_name.lower(), f"No monograph found for {drug_name}")

release_tools = [check_batch_purity, calculate_dosage_variance, lookup_drug_monograph]
release_agent_llm = get_llm(temperature=0).bind_tools(release_tools)

def batch_release_reasoning(state: MessagesState) -> dict:
    """The decision-making node: decides whether it has enough data to answer or needs a tool first."""
    response = release_agent_llm.invoke(state["messages"])
    return {"messages": [response]}

batch_release_tools_node = ToolNode(release_tools)

release_graph = StateGraph(MessagesState)
release_graph.add_node("batch_release_reasoning", batch_release_reasoning)
release_graph.add_node("batch_release_tools", batch_release_tools_node)
release_graph.add_edge(START, "batch_release_reasoning")
release_graph.add_conditional_edges("batch_release_reasoning", tools_condition,
                                     {"tools": "batch_release_tools", "__end__": END})
release_graph.add_edge("batch_release_tools", "batch_release_reasoning")

batch_release_app = release_graph.compile()

result = batch_release_app.invoke({
    "messages": [HumanMessage(content=(
        "Is batch PC-2026-0091 cleared for release? Check purity and confirm "
        "dosage variance for target 500mg, measured 512mg."
    ))]
})
print(result["messages"][-1].content)
````

# Example 4 — Manufacturing Supervisor

**Use case:** A single agent juggling purity checks, regulatory lookups,
and document drafting gets confused about which tool applies. PharmaOps
splits into specialists and adds a **Manufacturing Supervisor** that routes
the pharmacist's request to whichever specialist(s) can answer it.

````python
from typing import Literal, Annotated
from pydantic import BaseModel
from langgraph.graph.message import add_messages

def build_specialist_agent(tools_list, specialist_system_prompt):
    """Factory that produces a specialist node bound to its own tools and persona."""
    specialist_llm = get_llm(temperature=0).bind_tools(tools_list)

    def specialist_node(state) -> dict:
        messages = [SystemMessage(content=specialist_system_prompt)] + state["messages"]
        response = specialist_llm.invoke(messages)
        return {"messages": [response]}

    return specialist_node

quality_control_agent = build_specialist_agent(
    tools_list=[check_batch_purity, calculate_dosage_variance],
    specialist_system_prompt="""You are the Quality Control specialist. You
    verify batch purity and dosage variance against spec. Always check the
    actual lab data via tools before giving a verdict.""",
)

regulatory_agent = build_specialist_agent(
    tools_list=[lookup_drug_monograph],
    specialist_system_prompt="""You are the Regulatory Affairs specialist. You
    answer questions about drug monographs, classifications, and compliance
    requirements. Always consult the monograph tool before answering.""",
)

documentation_agent = build_specialist_agent(
    tools_list=[],
    specialist_system_prompt="""You are the Documentation specialist. You draft
    clear, GMP-compliant batch records and summaries from the information
    already gathered in this conversation.""",
)

class ManufacturingSupervisorState(TypedDict):
    messages: Annotated[list, add_messages]
    next_specialist: str

class SupervisorRoutingDecision(BaseModel):
    """The supervisor's decision about which specialist should act next."""
    reasoning: str
    next_specialist: Literal["quality_control_agent", "regulatory_agent",
                              "documentation_agent", "FINISH"]

def route_to_specialist(state: ManufacturingSupervisorState) -> dict:
    """Reads the conversation and decides which specialist should act next, or FINISH."""
    supervisor_routing_llm = get_llm(temperature=0).with_structured_output(SupervisorRoutingDecision)
    supervisor_prompt = """You coordinate three PharmaOps specialists:
    - quality_control_agent: purity checks, dosage variance
    - regulatory_agent: drug monographs, compliance classification
    - documentation_agent: drafts final batch documentation
    - FINISH: when the pharmacist's question is fully answered
    """
    messages = [SystemMessage(content=supervisor_prompt)] + state["messages"]
    decision = supervisor_routing_llm.invoke(messages)
    print(f"[SUPERVISOR] -> {decision.next_specialist} ({decision.reasoning})")
    return {"next_specialist": decision.next_specialist}

manufacturing_graph = StateGraph(ManufacturingSupervisorState)
manufacturing_graph.add_node("supervisor", route_to_specialist)
manufacturing_graph.add_node("quality_control_agent", quality_control_agent)
manufacturing_graph.add_node("regulatory_agent", regulatory_agent)
manufacturing_graph.add_node("documentation_agent", documentation_agent)
manufacturing_graph.add_node("qc_tools", ToolNode([check_batch_purity, calculate_dosage_variance]))
manufacturing_graph.add_node("regulatory_tools", ToolNode([lookup_drug_monograph]))

manufacturing_graph.add_edge(START, "supervisor")
manufacturing_graph.add_conditional_edges("supervisor", lambda s: s["next_specialist"], {
    "quality_control_agent": "quality_control_agent",
    "regulatory_agent": "regulatory_agent",
    "documentation_agent": "documentation_agent",
    "FINISH": END,
})
manufacturing_graph.add_conditional_edges("quality_control_agent", tools_condition,
                                           {"tools": "qc_tools", "__end__": "supervisor"})
manufacturing_graph.add_conditional_edges("regulatory_agent", tools_condition,
                                           {"tools": "regulatory_tools", "__end__": "supervisor"})
manufacturing_graph.add_edge("documentation_agent", "supervisor")
manufacturing_graph.add_edge("qc_tools", "quality_control_agent")
manufacturing_graph.add_edge("regulatory_tools", "regulatory_agent")

manufacturing_supervisor_app = manufacturing_graph.compile()
````

````python
result = manufacturing_supervisor_app.invoke({
    "messages": [HumanMessage(content=(
        "Is batch PC-2026-0091 within dosage tolerance for target 500mg measured 512mg, "
        "and what's the monograph classification for metformin?"
    ))],
    "next_specialist": "",
})
print("\nFINAL ANSWER:\n", result["messages"][-1].content)
````

# Example 5 — Formulation R&D Pipeline

**Use case:** Before PharmaCore reformulates a drug (e.g., to improve
bioavailability), R&D needs a structured literature review — plan the
sub-questions, search literature for each, synthesize gaps, write a report
the head chemist will actually read. This feeds new compounds back into
raw material intake.

````python
from typing import List
from operator import add

rd_llm = get_llm(temperature=0.3)

class FormulationResearchState(TypedDict):
    research_question: str
    research_subtasks: Annotated[List[str], add]
    literature_findings: Annotated[List[str], add]
    gap_analysis: str
    formulation_report: str

@tool
def search_pharma_literature(query: str) -> str:
    """Search pharmaceutical literature and patents for a formulation topic."""
    mock_results = {
        "bioavailability": "3 relevant papers on solubility enhancement via "
                            "nanocrystal formulation and lipid-based delivery.",
        "solubility": "2 papers on cyclodextrin complexation improving aqueous solubility.",
        "stability": "1 paper on polymorph screening to improve thermal stability.",
    }
    for keyword, result in mock_results.items():
        if keyword in query.lower():
            return result
    return f"Limited results for '{query}'."

def plan_formulation_research(state: FormulationResearchState) -> dict:
    """Breaks the formulation research question into 3-4 investigable subtasks."""
    prompt = f"""Break this formulation research question into 3-4 short, specific
    subtasks: {state['research_question']}. Return each on its own line, numbered."""
    response = rd_llm.invoke([HumanMessage(content=prompt)])
    subtasks = [line.strip() for line in response.content.strip().split("\n")
                if line.strip() and line.strip()[0].isdigit()]
    return {"research_subtasks": subtasks or [state["research_question"]]}

def search_formulation_literature(state: FormulationResearchState) -> dict:
    """Searches literature for each subtask the planner identified."""
    findings = []
    search_llm = rd_llm.bind_tools([search_pharma_literature])
    for subtask in state["research_subtasks"]:
        response = search_llm.invoke([HumanMessage(content=f"Research: {subtask}")])
        if getattr(response, "tool_calls", None):
            result = search_pharma_literature.invoke(response.tool_calls[0]["args"])
            findings.append(f"{subtask}\nFindings: {result}")
        else:
            findings.append(f"{subtask}\nFindings: {response.content}")
    return {"literature_findings": findings}

def analyze_formulation_gaps(state: FormulationResearchState) -> dict:
    """Synthesizes literature findings and identifies open formulation gaps."""
    findings_text = "\n\n".join(state["literature_findings"])
    prompt = f"Synthesize these findings and identify formulation gaps:\n{findings_text}"
    response = rd_llm.invoke([HumanMessage(content=prompt)])
    return {"gap_analysis": response.content}

def write_formulation_report(state: FormulationResearchState) -> dict:
    """Produces the final formulation R&D report for the head chemist."""
    prompt = f"""Write a short formulation R&D report for the head chemist.
    Question: {state['research_question']}
    Analysis: {state['gap_analysis']}
    Sections: Overview, Key Findings, Formulation Gaps, Recommended Next Steps."""
    response = rd_llm.invoke([HumanMessage(content=prompt)])
    return {"formulation_report": response.content}

rd_pipeline_graph = StateGraph(FormulationResearchState)
rd_pipeline_graph.add_node("plan_formulation_research", plan_formulation_research)
rd_pipeline_graph.add_node("search_formulation_literature", search_formulation_literature)
rd_pipeline_graph.add_node("analyze_formulation_gaps", analyze_formulation_gaps)
rd_pipeline_graph.add_node("write_formulation_report", write_formulation_report)

rd_pipeline_graph.add_edge(START, "plan_formulation_research")
rd_pipeline_graph.add_edge("plan_formulation_research", "search_formulation_literature")
rd_pipeline_graph.add_edge("search_formulation_literature", "analyze_formulation_gaps")
rd_pipeline_graph.add_edge("analyze_formulation_gaps", "write_formulation_report")
rd_pipeline_graph.add_edge("write_formulation_report", END)

formulation_rd_app = rd_pipeline_graph.compile()
````

````python
result = formulation_rd_app.invoke({
    "research_question": "How can we improve the bioavailability of Compound X?",
    "research_subtasks": [],
    "literature_findings": [],
    "gap_analysis": "",
    "formulation_report": "",
})
print(result["formulation_report"])
````

# Example 6 — Drug Info & Pricing Assistant

**Use case:** Patients and pharmacists ask two kinds of questions about a
drug once it's released to market: "what does it treat / how do I take it"
(needs retrieval over monograph text) and "how much does it cost" (needs a
structured price lookup). This chapter uses an in-memory vector store built
from a few sample monographs so it runs with no external files.

````python
import pandas as pd
from langchain_chroma import Chroma
from langchain.tools.retriever import create_retriever_tool
from langgraph.prebuilt import create_react_agent
from langgraph.checkpoint.memory import MemorySaver
from langchain_core.documents import Document

# In-memory pricing table (swap for your 650k-drug CSV in production)
drug_price_table = pd.DataFrame([
    {"drug_name": "Metformin 500mg", "price_usd": 4.50},
    {"drug_name": "Amoxicillin 250mg", "price_usd": 6.20},
    {"drug_name": "Atorvastatin 20mg", "price_usd": 9.85},
])

@tool
def get_drug_price(drug_name: str) -> str:
    """Look up the retail price of a drug by name (substring match)."""
    matches = drug_price_table[
        drug_price_table["drug_name"].str.contains(drug_name, case=False, na=False)
    ]
    if matches.empty:
        return f"No pricing found for '{drug_name}'."
    row = matches.iloc[0]
    return f"{row['drug_name']}: ${row['price_usd']:.2f}"

# In-memory monograph documents (swap for your real drug dataset in production)
drug_monograph_documents = [
    Document(page_content="Metformin: oral biguanide for type 2 diabetes. "
             "Typical dose 500-1000mg twice daily with meals. Store below 25C. "
             "Common side effect: GI upset."),
    Document(page_content="Amoxicillin: penicillin-class antibiotic. Typical dose "
             "250-500mg every 8 hours. Avoid in penicillin allergy."),
    Document(page_content="Atorvastatin: statin for cholesterol management. "
             "Typical dose 10-20mg once daily, evening administration preferred."),
]

drug_monograph_vectorstore = Chroma.from_documents(
    documents=drug_monograph_documents,
    embedding=get_embeddings(),
    collection_name="drug_monographs",
)
drug_info_retriever_tool = create_retriever_tool(
    drug_monograph_vectorstore.as_retriever(),
    name="get_drug_features",
    description="Retrieve indications, dosage form, and warnings for a drug from its monograph.",
)

drug_assistant_memory = MemorySaver()
drug_info_pricing_agent = create_react_agent(
    model=get_llm(temperature=0),
    tools=[get_drug_price, drug_info_retriever_tool],
    checkpointer=drug_assistant_memory,
)

patient_session = {"configurable": {"thread_id": "patient-4471"}}
````

````python
# First question establishes which drug we're talking about
r1 = drug_info_pricing_agent.invoke(
    {"messages": [HumanMessage(content="What is Metformin used for and what's the typical dose?")]},
    patient_session,
)
print("Patient: What is Metformin used for and what's the typical dose?")
print("Assistant:", r1["messages"][-1].content, "\n")

# Follow-up relies on thread_id memory to resolve "it" -> Metformin
r2 = drug_info_pricing_agent.invoke(
    {"messages": [HumanMessage(content="How much does it cost?")]},
    patient_session,
)
print("Patient: How much does it cost?")
print("Assistant:", r2["messages"][-1].content)
````

# Example 7 — Raw Material Inventory Agent

**Use case:** Before formulation can start, the system needs to check and
decrement raw material stock. Unlike Example 3's prebuilt ReAct loop,
PharmaOps builds this one manually because it needs both **read** and
**write** tools with explicit control over routing.

````python
raw_material_stock = pd.DataFrame([
    {"material_id": "RM-101", "quantity_kg": 250.0},
    {"material_id": "RM-204", "quantity_kg": 80.0},
])

@tool
def get_material_stock(material_id: str) -> str:
    """Query current stock level (kg) for a raw material by ID."""
    row = raw_material_stock[raw_material_stock["material_id"] == material_id]
    if row.empty:
        return f"No stock record for {material_id}"
    return f"{material_id}: {row.iloc[0]['quantity_kg']} kg available"

@tool
def consume_material(material_id: str, quantity_kg: float) -> str:
    """Deduct quantity_kg from a raw material's stock when a batch enters formulation."""
    idx = raw_material_stock.index[raw_material_stock["material_id"] == material_id]
    if len(idx) == 0:
        return f"No stock record for {material_id}"
    raw_material_stock.loc[idx, "quantity_kg"] -= quantity_kg
    remaining = raw_material_stock.loc[idx, "quantity_kg"].values[0]
    return f"Consumed {quantity_kg}kg of {material_id}. Remaining: {remaining}kg"

inventory_tools = [get_material_stock, consume_material]
inventory_agent_llm = get_llm(temperature=0).bind_tools(inventory_tools)

def inventory_agent_reasoning(state: MessagesState) -> dict:
    """Decides whether to query stock, consume stock, or answer directly."""
    return {"messages": [inventory_agent_llm.invoke(state["messages"])]}

def has_pending_tool_call(state: MessagesState) -> str:
    """Routes to the inventory tools node if the agent requested a tool call."""
    last_message = state["messages"][-1]
    return "inventory_tools" if getattr(last_message, "tool_calls", None) else END

class InventoryAgent:
    """Encapsulates the raw material inventory graph, LLM, and tools as one reusable unit."""

    def __init__(self):
        graph = StateGraph(MessagesState)
        graph.add_node("inventory_agent_reasoning", inventory_agent_reasoning)
        graph.add_node("inventory_tools", ToolNode(inventory_tools))
        graph.add_edge(START, "inventory_agent_reasoning")
        graph.add_conditional_edges("inventory_agent_reasoning", has_pending_tool_call,
                                     {"inventory_tools": "inventory_tools", END: END})
        graph.add_edge("inventory_tools", "inventory_agent_reasoning")
        self.app = graph.compile()

    def ask(self, message: str) -> str:
        result = self.app.invoke({"messages": [HumanMessage(content=message)]})
        return result["messages"][-1].content

inventory_agent = InventoryAgent()
print(inventory_agent.ask("How much RM-101 do we have, and consume 30kg of it for the next batch?"))
````

# Example 8 — Batch Deviation Report Agent

**Use case:** When a batch fails the QC checkpoint — off-spec purity, a
temperature excursion — GMP requires a formal Deviation Report with root
cause and corrective action (CAPA). PharmaOps uses a **generate -> review ->
revise** reflection loop so a QA reviewer's critique is baked in before a
human ever sees the draft.

````python
deviation_llm = get_llm(temperature=0.3)
MAX_REVISION_ROUNDS = 3

def draft_deviation_report(state: MessagesState) -> dict:
    """Generates or revises the batch deviation report based on incident notes and prior critique."""
    drafter_persona = SystemMessage(content="""
    You are drafting a GMP batch deviation report: Incident Summary, Root Cause,
    Impact Assessment, CAPA (Corrective and Preventive Action). Be specific and
    auditable.
    """)
    response = deviation_llm.invoke([drafter_persona] + state["messages"])
    return {"messages": [response]}

def review_deviation_report(state: MessagesState) -> dict:
    """Critiques the current draft against GMP documentation standards."""
    reviewer_persona = SystemMessage(content="""
    You are a QA reviewer. Critique the deviation report draft: is the root
    cause specific enough? Is the CAPA actionable? List concrete gaps, or say
    'APPROVED' if it meets GMP documentation standards.
    """)
    response = deviation_llm.invoke([reviewer_persona] + state["messages"])
    return {"messages": [response]}

def needs_another_revision(state: MessagesState) -> str:
    """Loops back to the drafter unless the reviewer approved or the round cap is hit."""
    revision_rounds = len(state["messages"])
    last_review = state["messages"][-1].content
    if "APPROVED" in last_review or revision_rounds >= (MAX_REVISION_ROUNDS * 2):
        return END
    return "draft_deviation_report"

deviation_report_graph = StateGraph(MessagesState)
deviation_report_graph.add_node("draft_deviation_report", draft_deviation_report)
deviation_report_graph.add_node("review_deviation_report", review_deviation_report)
deviation_report_graph.add_edge(START, "draft_deviation_report")
deviation_report_graph.add_edge("draft_deviation_report", "review_deviation_report")
deviation_report_graph.add_conditional_edges("review_deviation_report", needs_another_revision,
                                              {"draft_deviation_report": "draft_deviation_report", END: END})

deviation_report_app = deviation_report_graph.compile(checkpointer=MemorySaver())
````

````python
deviation_session = {"configurable": {"thread_id": "deviation-PC-2026-0044"}}
result = deviation_report_app.invoke(
    {"messages": [HumanMessage(content=(
        "Batch PC-2026-0044 failed QC: purity 96.1%, impurity profile exceeded threshold. "
        "Root cause suspected: mixing time under-run by 12 minutes on Line 2. Draft the deviation report."
    ))]},
    deviation_session,
)
print(result["messages"][-1].content)
````

# Example 9 — Plant Operations Router

**Use case:** A plant operator's chat interface shouldn't require them to
know which specialist agent to talk to. PharmaOps composes the **Drug Info
& Pricing Assistant** (Example 6) and the **Raw Material Inventory Agent**
(Example 7) behind one router that classifies intent and dispatches.

````python
from functools import partial

router_llm = get_llm(temperature=0)

class OperationsIntentClassification(BaseModel):
    """Classifies which subsystem an incoming query belongs to."""
    intent: Literal["DRUG_INFO", "INVENTORY", "SMALL_TALK", "UNCLEAR"]

def classify_operations_query(state: MessagesState) -> dict:
    """Reads the latest message and classifies it into a routing intent."""
    classification = router_llm.with_structured_output(OperationsIntentClassification).invoke(
        state["messages"]
    )
    return {"intent": classification.intent}

def run_drug_info_agent(state: MessagesState, agent) -> dict:
    """Delegates to the Drug Info & Pricing Assistant built in Example 6."""
    result = agent.invoke({"messages": state["messages"]}, patient_session)
    return {"messages": [result["messages"][-1]]}

def run_inventory_agent(state: MessagesState, inventory_agent: InventoryAgent) -> dict:
    """Delegates to the Raw Material Inventory Agent built in Example 7."""
    reply = inventory_agent.ask(state["messages"][-1].content)
    return {"messages": [HumanMessage(content=reply)]}

def handle_small_talk(state: MessagesState) -> dict:
    """Handles greetings and farewells without invoking a specialist."""
    response = router_llm.invoke(state["messages"])
    return {"messages": [response]}

class PlantOperationsState(MessagesState):
    intent: str

plant_ops_graph = StateGraph(PlantOperationsState)
plant_ops_graph.add_node("classify_operations_query", classify_operations_query)
plant_ops_graph.add_node("drug_info_agent", partial(run_drug_info_agent, agent=drug_info_pricing_agent))
plant_ops_graph.add_node("inventory_agent", partial(run_inventory_agent, inventory_agent=inventory_agent))
plant_ops_graph.add_node("small_talk", handle_small_talk)

plant_ops_graph.add_edge(START, "classify_operations_query")
plant_ops_graph.add_conditional_edges("classify_operations_query", lambda s: s["intent"], {
    "DRUG_INFO": "drug_info_agent",
    "INVENTORY": "inventory_agent",
    "SMALL_TALK": "small_talk",
    "UNCLEAR": END,
})
plant_ops_graph.add_edge("drug_info_agent", END)
plant_ops_graph.add_edge("inventory_agent", END)
plant_ops_graph.add_edge("small_talk", END)

plant_operations_app = plant_ops_graph.compile()

result = plant_operations_app.invoke({"messages": [HumanMessage(content="How much RM-204 do we have left?")]})
print(result["messages"][-1].content)
````

# Example 10 — Adverse Event Literature Screener

**Use case:** PharmaCore's pharmacovigilance team screens case reports for
a periodic safety review, applying an inclusion criterion that's easy to
misapply on a single pass (e.g., wrongly excluding pre-vetted regulatory-
database reports). The same reflection pattern from Example 8 applies here
at batch scale across a small in-memory set of reports.

````python
import re

screening_llm = get_llm(temperature=0)

INCLUSION_CRITERION = """
Exclude case reports lacking a causality assessment - EXCLUDING reports
sourced from recognized regulatory adverse-event databases (e.g., FAERS),
which are pre-vetted and should NOT be excluded on this basis.
"""

def screen_case_report(state: MessagesState) -> dict:
    """Applies the inclusion criterion to a single adverse event case report."""
    screener_persona = SystemMessage(content=f"""
    You are screening adverse event case reports for a safety review.
    Criterion: {INCLUSION_CRITERION}
    Respond with exactly: Decision: INCLUDE/EXCLUDE  Reason: <one sentence>
    """)
    response = screening_llm.invoke([screener_persona] + state["messages"])
    return {"messages": [response]}

def review_screening_decision(state: MessagesState) -> dict:
    """Checks whether the screening decision correctly applied the criterion's exception."""
    reviewer_persona = SystemMessage(content="""
    Check: did the screener correctly avoid excluding a FAERS-sourced report
    just for lacking a causality assessment? If the decision is correct, say
    'APPROVED'. Otherwise explain the error.
    """)
    response = screening_llm.invoke([reviewer_persona] + state["messages"])
    return {"messages": [response]}

def screening_needs_revision(state: MessagesState) -> str:
    return END if "APPROVED" in state["messages"][-1].content else "screen_case_report"

screener_graph = StateGraph(MessagesState)
screener_graph.add_node("screen_case_report", screen_case_report)
screener_graph.add_node("review_screening_decision", review_screening_decision)
screener_graph.add_edge(START, "screen_case_report")
screener_graph.add_edge("screen_case_report", "review_screening_decision")
screener_graph.add_conditional_edges("review_screening_decision", screening_needs_revision,
                                      {"screen_case_report": "screen_case_report", END: END})
adverse_event_screener_app = screener_graph.compile()

# Small in-memory set of case reports (swap for a CSV of hundreds in production)
sample_case_reports = pd.DataFrame([
    {"report_id": "AE-001", "source": "FAERS",
     "narrative": "Patient reported nausea after Metformin. No causality assessment on file."},
    {"report_id": "AE-002", "source": "Physician self-report",
     "narrative": "Patient reported dizziness. No causality assessment provided."},
])

def screen_all_case_reports(case_reports: pd.DataFrame) -> pd.DataFrame:
    """Runs the screener across every case report and returns decisions as a DataFrame."""
    results = []
    for _, report in case_reports.iterrows():
        report_source = report["source"]
        report_narrative = report["narrative"]
        outcome = adverse_event_screener_app.invoke({
            "messages": [HumanMessage(content=f"Source: {report_source}\n{report_narrative}")]
        })
        final_text = outcome["messages"][-1].content
        decision = re.search(r"Decision:\s*(INCLUDE|EXCLUDE)", final_text)
        reason = re.search(r"Reason:\s*(.+)", final_text)
        results.append({
            "report_id": report["report_id"],
            "decision": decision.group(1) if decision else "UNKNOWN",
            "reason": reason.group(1) if reason else final_text[:200],
        })
    return pd.DataFrame(results)

screening_results = screen_all_case_reports(sample_case_reports)
screening_results
````

# Example 11 — Regulatory Guidance Summarizer

**Use case:** Regulatory affairs tracks dozens of FDA/EMA guidance
documents per year. Each needs a structured summary — scope, binding
requirements, what PharmaOps must change to comply. This cell uses an
in-memory sample guidance text; swap `extract_guidance_text` for the PDF
version shown in the markdown note below to run it on real documents.

````python
summarizer_llm = get_llm(temperature=0)

def summarize_guidance_document(guidance_text: str) -> str:
    """Produces a structured compliance summary from raw guidance text."""
    summarizer_persona = SystemMessage(content="""
    Summarize this regulatory guidance document with exactly these sections:
    Scope, Key Requirements, Compliance Actions, Effective Date/Deadline.
    """)
    response = summarizer_llm.invoke([summarizer_persona, HumanMessage(content=guidance_text)])
    return response.content

sample_guidance_text = """
FDA Guidance for Industry: Process Validation for Solid Oral Dosage Forms (2026 update).
This guidance applies to manufacturers of solid oral dosage forms (tablets, capsules).
Manufacturers must demonstrate process validation across three stages: process design,
process qualification, and continued process verification. Continuous monitoring data
must be retained for a minimum of 7 years. Manufacturers have 18 months from publication
to update their validation master plans. Effective date: January 1, 2027.
"""

print(summarize_guidance_document(sample_guidance_text))
````

**To run this on real PDFs instead of the sample text above:**

```python
import requests
from pypdf import PdfReader

def download_guidance_pdf(guidance_id: str, url: str, cache_dir: str = "data/guidance_pdfs") -> str:
    os.makedirs(cache_dir, exist_ok=True)
    local_path = f"{cache_dir}/{guidance_id}.pdf"
    if not os.path.exists(local_path):
        response = requests.get(url, timeout=30)
        with open(local_path, "wb") as f:
            f.write(response.content)
    return local_path

def extract_guidance_text(pdf_path: str, max_pages: int = 20) -> str:
    reader = PdfReader(pdf_path)
    return "\n".join(page.extract_text() or "" for page in reader.pages[:max_pages])
```
